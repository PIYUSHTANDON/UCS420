```python
import pandas as pd
```

**Q1**

```python
roll_no = input("Enter your college roll number: ")
last_two_digits = [int(roll_no[-2]), int(roll_no[-1])]

categories = ["billing", "account", "general"]

fixed_entries = [
    {
        "question": "what is the annual fee",
        "answer": "The annual fee is Rs 500.",
        "keywords": "fee cost price charge",
        "category": "billing"
    },
    {
        "question": "how to reset password",
        "answer": "Go to Settings > Reset Password.",
        "keywords": "password reset login",
        "category": "account"
    },
    {
        "question": "what are your working hours",
        "answer": "We are open 9 AM to 5 PM.",
        "keywords": "hours timing open time",
        "category": "general"
    },
    {
        "question": "how can i pay the fee",
        "answer": "You can pay via UPI, card, or net banking.",
        "keywords": "pay payment upi fee",
        "category": "billing"
    }
]

personalized_entries = []

for digit in last_two_digits:
    category = categories[digit % 3]

    if category == "billing":
        question = "how can I update my billing details"
        answer = "Go to Billing Settings and update your billing details."
        keywords = "billing payment invoice update"

    elif category == "account":
        question = "how do I update my registered mobile number"
        answer = "Go to Account Settings > Personal Details and update your mobile number."
        keywords = "mobile number update phone account"

    else:
        question = "what are the general support options"
        answer = "You can contact support through the Help section."
        keywords = "support help general information"

    personalized_entries.append({
        "question": question,
        "answer": answer,
        "keywords": keywords,
        "category": category
    })

all_entries = fixed_entries + personalized_entries

df = pd.DataFrame(all_entries)

print("\nFinal DataFrame:")
print(df)
```

**Q2**

```python
def score_query(query, df):
    query_words = set(query.lower().split())
    results = []

    for _, row in df.iterrows():
        keywords = set(row["keywords"].lower().split())
        question_words = set(row["question"].lower().split())

        keyword_matches = query_words.intersection(keywords)
        question_matches = query_words.intersection(question_words)

        score = len(keyword_matches) + len(question_matches)

        if score > 0:
            results.append({
                "question": row["question"],
                "answer": row["answer"],
                "category": row["category"],
                "score": score
            })

    results = sorted(results, key=lambda x: x["score"], reverse=True)

    return results


query = input("Query: ")

results = score_query(query, df)

print("\nMatching :")
for result in results:
    print(result)
```

**Q6**

```python
def score_query_with_ties(query, df):
    query_words = set(query.lower().split())
    results = []

    for _, row in df.iterrows():

        keywords = set(row["keywords"].lower().split())
        question_words = set(row["question"].lower().split())

        keyword_matches = query_words.intersection(keywords)
        question_matches = query_words.intersection(question_words)

        score = len(keyword_matches) + len(question_matches)

        results.append({
            "question": row["question"],
            "answer": row["answer"],
            "category": row["category"],
            "score": score
        })

    highest_score = max(result["score"] for result in results)

    best_results = [
        result for result in results
        if result["score"] == highest_score
    ]

    return best_results


query = input("Enter query ")

results = score_query_with_ties(query, df)

print("\nmatching entries:")

for result in results:
    print(result)
```
