**3. Set the random seed to your own roll number using random.seed(your_roll_number) before generating your numbers — this keeps your list random-looking but reproducible and unique to you (a different seed produces a different list, so this cannot be copied from a classmate).**

**3.1 Generate a list of 100 random numbers between 100 and 900 (inclusive)**

```python
import random
random.seed(1024160081)
numbers = [random.randint(100,900) for _ in range(100)]
print(numbers)
```

**3.2  Count and print all odd numbers and even number  in the list**

```python
odd_nums=[num for num in numbers if num%2!=0]
print("Odd numbers:",odd_nums)

even_nums=[num for num in numbers if num%2==0]
print("Even numbers:",even_nums)
```

**3.4 Count and print all prime numbers in the list, and additionally build the actual list of prime numbers found using list comprehension.**

```python
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True
prime_numbers = [x for x in numbers if is_prime(x)]
print("Prime numbers:",prime_numbers)
print("Number of prime numbers:", len(prime_numbers))
```

**3.5 Print the number that occurs most frequently in your list, and how many times it occurs**

```python
frequency = {}
for i in numbers:
    frequency[i] = frequency.get(i, 0) + 1
most_frequent = max(frequency, key=frequency.get)
most_frequent_count = frequency[most_frequent]
print("Most frequent number is: ", most_frequent, "with count: ", most_frequent_count)
```

**4. Using the same 8 digits of your roll number from Q1 (before multiplying by 10), create set A = {digit × 7 for each digit} and set B = {digit × 9 for each digit}. Print both sets.**

**4.1 Find the union of A and B — all unique values across both sets. and also  find the intersection of A and B — values common to both.**

```python
roll_no = "1024160081"
digits = [int(digit) for digit in roll_no[:8]]
A = {digit * 7 for digit in digits}
B = {digit * 9 for digit in digits}
print("Set A:",A)
print("Set B:",B)

print("Union",A.union(B))

print("Intersection",A.intersection(B))
```
